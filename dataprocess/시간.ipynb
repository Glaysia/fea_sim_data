```python
import pandas as pd

# 1. Load the dataset (adjust filename/path if needed)
df = pd.read_csv('data_분류된_형상설계.csv')

# 2. Define success: fail_level == 15
df['is_success'] = (df['fail_level'] == 15)

# 3. Overall elapsed time statistics
print("Overall elapsed time statistics:")
print(df['elapsed'].describe(), end="\n\n")

# 4. Counts of success vs. failure
counts = df['is_success'].value_counts()
print("Counts of success (True) vs. failure (False):")
print(counts, end="\n\n")

# 5. Success rate
success_rate = counts[True] / counts.sum()
print(f"Success rate: {success_rate:.4f}", end="\n\n")

# 6. Elapsed time statistics by success/failure
print("Elapsed time statistics by group:")
print(df.groupby('is_success')['elapsed'].describe())

# 7. (Optional) Save summary to CSV
summary = df.groupby('is_success')['elapsed'].describe().round(2)
summary.to_csv('elapsed_time_stats_by_success.csv', float_format='%.2f')
print("\nSaved group statistics to 'elapsed_time_stats_by_success.csv'.")
```

```python
import pandas as pd

# Load dataset
df = pd.read_csv('data_분류된_형상설계.csv')

# Mark success where fail_level == 15
df['is_success'] = (df['fail_level'] == 15)

# Compute total elapsed time per group
total_times = df.groupby('is_success')['elapsed'].sum()

# Extract values
total_fail_time    = total_times.get(False, 0)
total_success_time = total_times.get(True, 0)

print(f"Total elapsed time for failures: {total_fail_time/1e9/3600}")
print(f"Total elapsed time for successes: {total_success_time/1e9/3600}")
```

```python
import pandas as pd

df = pd.read_csv('data_분류된_형상설계.csv')
df['new']   = (df['g2'] > (df['Tx_height'] + df['Tx_preg'] / 2)).astype(int)
df['is_15'] = (df['fail_level'] == 15).astype(int)

# 집계
summary = df.groupby('new')['is_15'].agg(['count','sum'])
summary['success_rate'] = (summary['sum'] / summary['count']).round(4)

# new==0, new==1, 전체
total = pd.DataFrame({
    'count': [df.shape[0]],
    'sum':   [df['is_15'].sum()]
})
total['success_rate'] = (total['sum'] / total['count']).round(4)
total.index = ['All']

result = pd.concat([summary, total])
print(result)
```
