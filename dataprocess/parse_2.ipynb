```python
import os, glob, pathlib, json
# f = open("../.gitignore", "r")

data_dir = pathlib.Path('../tmp').resolve()
data_dirs = os.listdir(data_dir)

dicts_list = []

for computer_name in data_dirs:
    for fp in glob.glob(f'{str(data_dir)}/{computer_name}/*.json'):
        try:
            with open(fp, 'r', encoding='utf-8') as f:
                _dict = json.load(f)
                uuid = f.name
                start = uuid.index('sim_dict_') + len('sim_dict_')
                uuid = uuid[start:]
                uuid = uuid.removesuffix(".json")
                _dict['uuid'] = uuid
                _dict['computer_name'] = computer_name

                dicts_list.append(_dict)
        except json.JSONDecodeError:
            print(f"Warning: failed to parse {fp}")

dicts_list
```

```python
len(dicts_list)
```

```python
new = []
fail =[]
for i in dicts_list:
    dd = i['progress_dict']
    go = dd[list(dd).pop()][0] =='coreloss_project'
    # print(dd[list(dd).pop()][0])
    if go:
        new.append(i)
    else:
        fail.append(i)
len(new)
```

```python
len(fail)
```

```python
import pandas as pd
df = pd.DataFrame(new)
df.head().columns
variables = df['v']
data = df['data']
# data[0].keys()
computer_name = df['computer_name']
df[['uuid']]
df[['computer_name']]
# computer_name
```

```python
"v"
variables_df = pd.DataFrame(variables.to_list())

variables_df.drop(
    columns=[
    'seed', 'A', 'B', 'C', 'D',
    'E'
    ], inplace=True
    )

variables_df.head()
```

```python
"_get_copper_loss_parameter"
"_get_copper_loss_parameter_after"
data_df = pd.DataFrame(data.to_list())
copper_1 = data_df.pop("_get_copper_loss_parameter")
copper_2 = data_df.pop("_get_copper_loss_parameter_after")

flattened1 = copper_1.map(lambda d: {k: list(v.values())[0] for k, v in d.items()})
flattened2 = copper_2.map(lambda d: {k: list(v.values())[0] for k, v in d.items()})

# DataFrame으로 변환
copper_1_df = pd.DataFrame(flattened1.tolist())
copper_2_df = pd.DataFrame(flattened2.tolist())

copper_1_df.columns = ['P1_Tx [W]', 'P1_Rx1 [W]', 'P1_Rx2 [W]']
copper_2_df.columns = ['P2_Tx [W]', 'P2_Rx1 [W]', 'P2_Rx2 [W]']
copper_1_df.head()
copper_2_df.head()

copper_df = copper_1_df.join(copper_2_df, how='left')

copper_df.head()
```

```python
"__get_B_field" # B field [T]
data_df = pd.DataFrame(data.to_list())
b_field = data_df.pop("__get_B_field")
b_field = pd.DataFrame(b_field.tolist())

b_field.columns = [
    'B_mean_left [T]', 'B_mean_center [T]', 'B_mean_right [T]',
    'B_mean_top_left [T]', 'B_mean_bottom_left [T]',
    'B_mean_top_right [T]', 'B_mean_bottom_right [T]'
    ]
b_field.head()
```

```python
"_get_magnetic_report"
data_df = pd.DataFrame(data.to_list())
magnetic = pd.DataFrame(data_df.pop('_get_magnetic_report').to_list())
magnetic.columns = [
    'Ltx_uH', 'Lrx1_uH', 'Lrx2_uH', 'M1', 'M2', 'k1', 'k2',
    'Lmt_uH', 'Lmr1_uH', 'Lmr2_uH', 'Llt_uH', 'Llr1_uH', 'Llr2_uH'
]

magnetic.head()
```

```python

"magnetizing_current"

data_df = pd.DataFrame(data.to_list())
mag_cur = pd.DataFrame(data_df.pop('magnetizing_current').to_list())

mag_cur.columns = ['Magnetizing Current [A]']
mag_cur.head()
```

```python
'coreloss_project'
data_df = pd.DataFrame(data.to_list())

coreloss = pd.DataFrame(data_df.pop('coreloss_project').to_list())


# 1) Coreloss [mW]→W, [kW]→W 로 변환해서 우선 채우고, 마지막에 [W] 로 채움
coreloss['Coreloss'] = (
    coreloss['Coreloss [W]']
      .fillna(coreloss.get('Coreloss [mW]', 0) * 1e-3)
      .fillna(coreloss.get('Coreloss [kW]', 0) * 1e3)
)

# 2) 이제 단위별 원본 컬럼들 삭제
coreloss.drop(columns=[c for c in coreloss.columns if c.startswith('Coreloss [')], inplace=True)
coreloss.columns = ['Coreloss_W']
coreloss
```

```python
new_data_df = copper_df.join(b_field, how='left')
new_data_df = new_data_df.join(magnetic, how='left')
new_data_df = new_data_df.join(mag_cur, how='left')
new_data_df = new_data_df.join(coreloss, how='left')
```

```python

df_new = df[['uuid']]
df_new = df_new.join(df[['computer_name']])
df_new = df_new.join(variables_df)
df_new = df_new.join(new_data_df)

df_new.head()
```
